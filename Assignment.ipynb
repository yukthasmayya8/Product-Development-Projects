```python
"""1b. Create a function calc1(n1, n2) to take the 2 input values and print the output Assignment 1 - Create Mathematical Tables from 3 to 20"""

def calc1(n1, n2):
    sum1 = n1 + n2
    dif1 = n1 - n2
    prd1 = n1 * n2
    div1 = n1 / n2
    div2 = n2 / n1
    rem1 = n1 % n2
    exp1 = n1 ** n2
    print(sum1, dif1, prd1, div1, div2, rem1, exp1)

n1 = int(input("Enter first number: "))
n2 = int(input("Enter second number: "))
calc1(n1, n2)
```

```python
"""2c. Take input from a file in1.txt containing 3 in first line and 8 in second line and generate the mathematical tables 3
8"""

with open("in1.txt", 'w') as ofile:
    ofile.write("3")
    ofile.write("\n8")

list1 = []
with open("in1.txt", 'r') as infile:
    for v in infile:
        s = v.split()
        for i in s:
            list1.append(int(i))

for i in list1:
    for j in range(1,11):
        print(i," * ",j," = ", j * i)
    print("\n")
```

```python
"""2d. Take input from a file in2.txt containing 3,8 in the first line and generate the mathematical tables
3,8"""

with open("in2.txt", 'w') as ofile:
    ofile.write("3,")
    ofile.write("8")

list1=[]
with open("in2.txt",'r') as infile:
    for v in infile:
        s = v.split(",")
        for i in s:
            list1.append(int(i))
for i in list1:
    for j in range(1,11):
        print(i," * ",j ," = ", j * 1)
    print("\n")
```

```python
"""2e. Send the results to an output file out1.txt. Use "\n" wherever required to generate a new line. The file content should look same as the console output"""

with open("in1.txt", 'w') as ofile:
    ofile.write("3")
    ofile.write("\n8")

list1 = []

with open("in1.txt", 'r') as infile:
    for v in infile:
        s = v.split()
        for i in s:
            list1.append(int(i))

for i in list1:
    for j in range(1,11):
        print(i, " * ",j," = ", j * i)
    print("\n")

with open("out1.txt", "w") as outfile:
    for i in list1:
        for j in range(1,11):
            outfile.write(f"{i} * {j} = {i*j}")
            outfile.write("\n")
        outfile.write("\n")
```

```python
"""2f. Send the results to 6 different output files 3.txt, 4.txt, 5.txt, 6.txt, 7.txt, 8.txt depending on the start and end values given as input. ie. The output filename should be dynamically created for each and every mathematical table generated"""

start = int(input("Enter first number: "))
end = int(input("Enter second number: "))

for i in range(start, end+1):
    filename = "out" + str(i+1) + ".txt"
    print(filename)

    with open(filename, "w") as outfile:
        for j in range(1,11):
            print(i," * ", j," = ", j * i)
            outfile.write(f"{i} * {j} = {i*j}")
            outfile.write("\n")
            outfile.write("\n")
    print("\n")
```
